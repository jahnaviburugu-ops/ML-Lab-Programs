```python
"""
Backpropagation Neural Network Prediction and Feature Analysis System
Dataset : Breast Cancer Wisconsin (Diagnostic)  -- 569 samples, 30 features
Classes : Benign = 0, Malignant = 1
Networks: BPNN, MLP, SHLNN  (all implemented from scratch with NumPy,
          sigmoid activations + backpropagation / gradient descent)

Data source (tried in this order):
  1. A local Kaggle/UCI CSV named  data.csv  (columns: id, diagnosis, 30 features)
  2. scikit-learn's built-in copy of the same WDBC dataset
"""
import os
import sys
import time
import numpy as np

SEED = 42
THRESHOLD = 0.5

# Confidence bands based on |output - 0.5|  (max possible distance = 0.5)
LOW_BAND = 0.15       # distance <  0.15  -> Low      (output in 0.35 .. 0.65)
HIGH_BAND = 0.35      # distance >= 0.35  -> High     (output <= 0.15 or >= 0.85)


# ----------------------------------------------------------------------------
# Data handling
# ----------------------------------------------------------------------------
def load_dataset():
    """Return X (569x30), y (Benign=0, Malignant=1), feature_names."""
    for path in ("data.csv", "wdbc.csv", "breast-cancer-wisconsin-data.csv"):
        if os.path.exists(path):
            import csv
            with open(path, newline="") as f:
                rows = list(csv.reader(f))
            header, body = rows[0], rows[1:]
            d_idx = header.index("diagnosis")
            feat_idx = [i for i, h in enumerate(header)
                        if h not in ("id", "diagnosis") and h.strip() != ""
                        and not h.startswith("Unnamed")]
            X = np.array([[float(r[i]) for i in feat_idx] for r in body if r])
            y = np.array([1 if r[d_idx] == "M" else 0 for r in body if r])
            names = [header[i] for i in feat_idx]
            print(f"[data] Loaded {path}")
            return X, y, names
    try:
        from sklearn.datasets import load_breast_cancer
    except ImportError:
        sys.exit("Place the Kaggle 'data.csv' next to this script, or "
                 "install scikit-learn (pip install scikit-learn).")
    ds = load_breast_cancer()
    y = 1 - ds.target            # sklearn: 0=malignant,1=benign -> flip to Benign=0, Malignant=1
    print("[data] Loaded WDBC dataset from scikit-learn")
    return ds.data.astype(float), y.astype(int), list(ds.feature_names)


def stratified_split(X, y, test_frac=0.2, seed=SEED):
    rng = np.random.default_rng(seed)
    tr, te = [], []
    for c in (0, 1):
        idx = rng.permutation(np.where(y == c)[0])
        n_te = int(round(len(idx) * test_frac))
        te += list(idx[:n_te])
        tr += list(idx[n_te:])
    tr, te = rng.permutation(tr), rng.permutation(te)
    return tr, te


# ----------------------------------------------------------------------------
# Neural network (fully-connected, sigmoid everywhere, trained by backprop)
# ----------------------------------------------------------------------------
def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-np.clip(z, -500, 500)))


class NeuralNetwork:
    def __init__(self, layers, lr=0.1, epochs=200, batch_size=None,
                 momentum=0.0, seed=SEED, name="NN"):
        self.layers, self.lr, self.epochs = layers, lr, epochs
        self.batch_size, self.momentum, self.name = batch_size, momentum, name
        rng = np.random.default_rng(seed)
        self.rng = rng
        self.W, self.b = [], []
        for n_in, n_out in zip(layers[:-1], layers[1:]):
            limit = np.sqrt(6.0 / (n_in + n_out))            # Xavier/Glorot init
            self.W.append(rng.uniform(-limit, limit, (n_in, n_out)))
            self.b.append(np.zeros(n_out))
        self.loss_history = []
        self.train_time = 0.0

    # forward pass: returns list of activations [input, h1, ..., output]
    def forward(self, X):
        acts = [X]
        for W, b in zip(self.W, self.b):
            acts.append(sigmoid(acts[-1] @ W + b))
        return acts

    def predict_proba(self, X):
        return self.forward(X)[-1].ravel()

    def predict(self, X):
        return (self.predict_proba(X) >= THRESHOLD).astype(int)

    def hidden_activations(self, X):
        """Activations of the (first) hidden layer."""
        return self.forward(X)[1]

    def fit(self, X, y):
        t0 = time.time()
        y = y.reshape(-1, 1).astype(float)
        n = len(X)
        bs = self.batch_size or n
        vW = [np.zeros_like(w) for w in self.W]
        vb = [np.zeros_like(b) for b in self.b]
        for _ in range(self.epochs):
            order = self.rng.permutation(n)
            for s in range(0, n, bs):
                idx = order[s:s + bs]
                xb, yb = X[idx], y[idx]
                m = len(idx)
                acts = self.forward(xb)
                # output delta (sigmoid + cross-entropy)  ->  (a - y)
                delta = (acts[-1] - yb) / m
                for l in range(len(self.W) - 1, -1, -1):
                    gW = acts[l].T @ delta
                    gb = delta.sum(axis=0)
                    if l > 0:   # propagate error back through sigmoid derivative
                        delta = (delta @ self.W[l].T) * acts[l] * (1 - acts[l])
                    vW[l] = self.momentum * vW[l] - self.lr * gW
                    vb[l] = self.momentum * vb[l] - self.lr * gb
                    self.W[l] += vW[l]
                    self.b[l] += vb[l]
            p = np.clip(self.predict_proba(X), 1e-12, 1 - 1e-12)
            self.loss_history.append(
                float(-np.mean(y.ravel() * np.log(p) + (1 - y.ravel()) * np.log(1 - p))))
        self.train_time = time.time() - t0
        return self

    def n_params(self):
        return sum(w.size for w in self.W) + sum(b.size for b in self.b)


# ----------------------------------------------------------------------------
# Helpers
# ----------------------------------------------------------------------------
CLASS_NAME = {0: "Benign", 1: "Malignant"}


def confidence_category(out):
    d = abs(out - THRESHOLD)
    if d < LOW_BAND:
        return "Low Confidence"
    if d < HIGH_BAND:
        return "Moderate Confidence"
    return "High Confidence"


def metrics(y_true, y_pred):
    tp = int(np.sum((y_true == 1) & (y_pred == 1)))
    tn = int(np.sum((y_true == 0) & (y_pred == 0)))
    fp = int(np.sum((y_true == 0) & (y_pred == 1)))
    fn = int(np.sum((y_true == 1) & (y_pred == 0)))
    acc = (tp + tn) / len(y_true)
    prec = tp / (tp + fp) if tp + fp else 0.0
    rec = tp / (tp + fn) if tp + fn else 0.0
    f1 = 2 * prec * rec / (prec + rec) if prec + rec else 0.0
    return dict(acc=acc, prec=prec, rec=rec, f1=f1, tp=tp, tn=tn, fp=fp, fn=fn)


def header(title):
    print("\n" + "=" * 78)
    print(title)
    print("=" * 78)


# ----------------------------------------------------------------------------
# The system
# ----------------------------------------------------------------------------
class System:
    def __init__(self):
        X, y, names = load_dataset()
        self.names = names
        tr, te = stratified_split(X, y)
        # Min-max normalisation (fitted on training data only -> no leakage)
        self.xmin, self.xmax = X[tr].min(axis=0), X[tr].max(axis=0)
        Xn = (X - self.xmin) / (self.xmax - self.xmin + 1e-12)
        self.Xtr, self.ytr = Xn[tr], y[tr]
        self.Xte, self.yte = Xn[te], y[te]
        self.test_ids = te
        print(f"[data] {len(y)} samples, {X.shape[1]} features | "
              f"Benign={int((y == 0).sum())}, Malignant={int((y == 1).sum())}")
        print(f"[data] Train = {len(tr)}, Test = {len(te)}  (stratified 80/20), "
              f"features normalised to [0,1]")
        self.models = {}

    # -- model factory (trained lazily, then cached) --------------------------
    def get(self, key):
        if key not in self.models:
            cfg = {
                # classic BPNN: one hidden layer, full-batch gradient descent + momentum
                "BPNN": dict(layers=[30, 10, 1], lr=0.5, epochs=800,
                             batch_size=None, momentum=0.9),
                # MLP: two hidden layers, mini-batch gradient descent
                "MLP": dict(layers=[30, 16, 8, 1], lr=0.1, epochs=200,
                            batch_size=32, momentum=0.0),
                # SHLNN: single hidden layer, mini-batch gradient descent
                "SHLNN": dict(layers=[30, 12, 1], lr=0.1, epochs=200,
                              batch_size=16, momentum=0.0),
            }[key]
            print(f"[train] Training {key} {cfg['layers']} ...", end=" ", flush=True)
            m = NeuralNetwork(name=key, **cfg).fit(self.Xtr, self.ytr)
            self.models[key] = m
            acc = np.mean(m.predict(self.Xte) == self.yte)
            print(f"done in {m.train_time:.2f}s (test accuracy {acc*100:.2f}%)")
        return self.models[key]

    # -- 1 --------------------------------------------------------------------
    def tumor_classification(self):
        header("1. TUMOR CLASSIFICATION USING BPNN  (30-10-1, sigmoid, backprop)")
        m = self.get("BPNN")
        out = m.predict_proba(self.Xte)
        pred = (out >= THRESHOLD).astype(int)
        print(f"{'Sample':>6} {'Actual':>10} {'Predicted':>10} {'Output':>9}  Result")
        print("-" * 52)
        for i in range(len(out)):
            ok = "OK" if pred[i] == self.yte[i] else "WRONG"
            print(f"{i+1:>6} {CLASS_NAME[self.yte[i]]:>10} {CLASS_NAME[pred[i]]:>10} "
                  f"{out[i]:>9.4f}  {ok}")
        mt = metrics(self.yte, pred)
        print("-" * 52)
        print(f"Correctly classified : {mt['tp']+mt['tn']} / {len(out)}")
        print(f"Classification accuracy: {mt['acc']*100:.2f}%")
        print(f"Final training loss    : {m.loss_history[-1]:.5f}")

    # -- 2 --------------------------------------------------------------------
    def confidence_analyzer(self):
        header("2. PREDICTION CONFIDENCE ANALYZER  (MLP 30-16-8-1)")
        print(f"Distance d = |output - 0.5|:  d < {LOW_BAND}: Low | "
              f"{LOW_BAND} <= d < {HIGH_BAND}: Moderate | d >= {HIGH_BAND}: High\n")
        m = self.get("MLP")
        out = m.predict_proba(self.Xte)
        pred = (out >= THRESHOLD).astype(int)
        counts = {"Low Confidence": 0, "Moderate Confidence": 0, "High Confidence": 0}
        print(f"{'Sample':>6} {'Output':>9} {'Predicted':>10}  Confidence")
        print("-" * 50)
        for i, o in enumerate(out):
            c = confidence_category(o)
            counts[c] += 1
            print(f"{i+1:>6} {o:>9.4f} {CLASS_NAME[pred[i]]:>10}  {c}")
        print("-" * 50)
        for k, v in counts.items():
            print(f"{k:<20}: {v:>3} ({v/len(out)*100:.1f}%)")

    # -- 3 --------------------------------------------------------------------
    def feature_contribution(self):
        header("3. FEATURE CONTRIBUTION ANALYZER  (MLP)")
        m = self.get("MLP")
        try:
            s = input(f"Test sample number (1-{len(self.Xte)}) [Enter = 1]: ").strip()
            idx = int(s) - 1 if s else 0
        except (ValueError, EOFError):
            idx = 0
        if not 0 <= idx < len(self.Xte):
            print("Invalid sample number, using sample 1.")
            idx = 0
        x = self.Xte[idx].copy()
        base = float(m.predict_proba(x[None, :])[0])
        print(f"\nSelected test sample #{idx+1}: actual = {CLASS_NAME[self.yte[idx]]}, "
              f"baseline output = {base:.4f} -> {CLASS_NAME[int(base >= THRESHOLD)]}")
        step = 0.25
        print(f"Each feature is perturbed by +/-{step} (normalised units, clipped to [0,1]); "
              f"influence = mean |change in output|.\n")
        rows = []
        for j in range(len(x)):
            ch = []
            for d in (+step, -step):
                xp = x.copy()
                xp[j] = np.clip(xp[j] + d, 0.0, 1.0)
                ch.append(float(m.predict_proba(xp[None, :])[0]) - base)
            rows.append((j, ch[0], ch[1], (abs(ch[0]) + abs(ch[1])) / 2))
        rows.sort(key=lambda r: r[3], reverse=True)
        print(f"{'Rank':>4}  {'Feature':<26} {'d(out) +':>10} {'d(out) -':>10} {'Influence':>10}")
        print("-" * 68)
        for r, (j, up, dn, inf) in enumerate(rows, 1):
            print(f"{r:>4}  {self.names[j]:<26} {up:>+10.4f} {dn:>+10.4f} {inf:>10.4f}")
        print("\nTop 5 most influential features:")
        for r, (j, _, _, inf) in enumerate(rows[:5], 1):
            print(f"  {r}. {self.names[j]}  (influence = {inf:.4f})")

    # -- 4 --------------------------------------------------------------------
    def hidden_neuron_response(self):
        header("4. HIDDEN NEURON RESPONSE ANALYZER  (SHLNN 30-12-1)")
        m = self.get("SHLNN")
        H = m.hidden_activations(self.Xte)
        n_h = H.shape[1]
        print("Hidden neuron activation matrix (rows = test samples, columns = hidden neurons)")
        print(f"{'Sample':>6} {'Class':>4} " + " ".join(f"{'H'+str(k+1):>6}" for k in range(n_h)))
        print("-" * (13 + 7 * n_h))
        for i in range(len(H)):
            print(f"{i+1:>6} {'M' if self.yte[i] else 'B':>4} " +
                  " ".join(f"{v:>6.3f}" for v in H[i]))
        mean_b = H[self.yte == 0].mean(axis=0)
        mean_m = H[self.yte == 1].mean(axis=0)
        print("\nAverage activation per hidden neuron")
        print(f"{'Neuron':>8} {'Benign':>9} {'Malignant':>10} {'Diff(M-B)':>10}")
        for k in range(n_h):
            print(f"{'H'+str(k+1):>8} {mean_b[k]:>9.4f} {mean_m[k]:>10.4f} {mean_m[k]-mean_b[k]:>+10.4f}")
        kb, km = int(np.argmax(mean_b)), int(np.argmax(mean_m))
        print(f"\nHighest average activation for BENIGN    : H{kb+1} ({mean_b[kb]:.4f})")
        print(f"Highest average activation for MALIGNANT : H{km+1} ({mean_m[km]:.4f})")

    # -- 5 --------------------------------------------------------------------
    def feature_masking(self):
        header("5. INPUT FEATURE MASKING CHALLENGE  (BPNN)")
        m = self.get("BPNN")
        base = np.mean(m.predict(self.Xte) == self.yte)
        mean_vec = self.Xtr.mean(axis=0)           # dataset mean (normalised space)
        print(f"Baseline accuracy (no masking): {base*100:.2f}%")
        print("Masking = replace one feature at a time by 0 (normalised minimum) "
              "or by its mean value.\n")
        res = []
        for j in range(self.Xte.shape[1]):
            r = [j]
            for mode in ("zero", "mean"):
                Xm = self.Xte.copy()
                Xm[:, j] = 0.0 if mode == "zero" else mean_vec[j]
                acc = np.mean(m.predict(Xm) == self.yte)
                r += [acc, base - acc]
            res.append(r)
        print(f"{'Feature':<26} {'Acc(zero)':>9} {'Drop':>7} {'Acc(mean)':>10} {'Drop':>7}")
        print("-" * 64)
        for j, az, dz, am, dm in res:
            print(f"{self.names[j]:<26} {az*100:>8.2f}% {dz*100:>6.2f}% "
                  f"{am*100:>9.2f}% {dm*100:>6.2f}%")
        print("-" * 64)
        for label, col in (("zero", 2), ("mean", 4)):
            best = max(res, key=lambda r: r[col])
            tied = [self.names[r[0]] for r in res if abs(r[col] - best[col]) < 1e-12]
            print(f"Max drop with {label}-masking: {best[col]*100:.2f}% "
                  f"-> {', '.join(tied[:5])}{' ...' if len(tied) > 5 else ''}")
        if all(r[2] <= 0 and r[4] <= 0 for r in res):
            print("(No single feature removal hurts accuracy -- the 30 features are highly redundant.)")

    # -- 6 --------------------------------------------------------------------
    def misclassification_analyzer(self):
        header("6. MISCLASSIFICATION CONFIDENCE ANALYZER  (MLP)")
        m = self.get("MLP")
        out = m.predict_proba(self.Xte)
        pred = (out >= THRESHOLD).astype(int)
        wrong = np.where(pred != self.yte)[0]
        print(f"Misclassified test samples: {len(wrong)} of {len(out)}")
        if len(wrong) == 0:
            print("The network classified every test sample correctly - nothing to analyse.")
            return
        print(f"\n{'Sample':>6} {'Actual':>10} {'Predicted':>10} {'Output':>9}  Confidence")
        print("-" * 62)
        confident = []
        for i in wrong:
            c = confidence_category(out[i])
            if c == "High Confidence":
                confident.append(i)
            print(f"{i+1:>6} {CLASS_NAME[self.yte[i]]:>10} {CLASS_NAME[pred[i]]:>10} "
                  f"{out[i]:>9.4f}  {c}")
        print("-" * 62)
        if confident:
            print(f"CONFIDENTLY WRONG predictions (output far from 0.5, |out-0.5| >= {HIGH_BAND}): "
                  f"{len(confident)} -> samples {[int(i)+1 for i in confident]}")
            fn = sum(1 for i in confident if self.yte[i] == 1)
            print(f"  of which {fn} are missed malignant tumours (false negatives) -- the "
                  f"clinically most dangerous error.")
        else:
            print("No confidently wrong predictions: every misclassified sample lies close to "
                  "the threshold,\nso the network 'knows' it is unsure on its errors.")

    # -- 7 --------------------------------------------------------------------
    def comparative_analysis(self):
        header("7. BPNN - MLP - SHLNN COMPARATIVE ANALYSIS")
        rows = []
        for k in ("BPNN", "MLP", "SHLNN"):
            m = self.get(k)
            pr_tr, pr_te = m.predict(self.Xtr), m.predict(self.Xte)
            mt = metrics(self.yte, pr_te)
            rows.append((k, m, np.mean(pr_tr == self.ytr), mt))
        print("\nArchitecture / training setup")
        print(f"{'Model':<7} {'Layers':<14} {'LR':>5} {'Epochs':>7} {'Batch':>6} {'Mom.':>5} {'Params':>7}")
        for k, m, _, _ in rows:
            print(f"{k:<7} {str(m.layers):<14} {m.lr:>5} {m.epochs:>7} "
                  f"{(m.batch_size or 'full'):>6} {m.momentum:>5} {m.n_params():>7}")
        print("\nPerformance on the test set")
        print(f"{'Model':<7} {'TrainAcc':>8} {'TestAcc':>8} {'Precis.':>8} {'Recall':>7} "
              f"{'F1':>6} {'FinalLoss':>10} {'Time(s)':>8}")
        for k, m, tra, mt in rows:
            print(f"{k:<7} {tra*100:>7.2f}% {mt['acc']*100:>7.2f}% {mt['prec']*100:>7.2f}% "
                  f"{mt['rec']*100:>6.2f}% {mt['f1']:>6.3f} {m.loss_history[-1]:>10.5f} "
                  f"{m.train_time:>8.2f}")
        print("\nConfusion matrices (rows = actual, cols = predicted; order Benign, Malignant)")
        for k, _, _, mt in rows:
            print(f"  {k:<6}  [[TN={mt['tn']:>3}, FP={mt['fp']:>3}],  [FN={mt['fn']:>3}, TP={mt['tp']:>3}]]")
        best = max(rows, key=lambda r: (r[3]['acc'], r[3]['rec']))
        print(f"\nBest test accuracy: {best[0]} ({best[3]['acc']*100:.2f}%)")
        print("Note: differences of 1-2 samples on a 113-sample test set are within noise.")


def main():
    system = System()
    actions = {
        "1": system.tumor_classification,
        "2": system.confidence_analyzer,
        "3": system.feature_contribution,
        "4": system.hidden_neuron_response,
        "5": system.feature_masking,
        "6": system.misclassification_analyzer,
        "7": system.comparative_analysis,
    }
    while True:
        print("\n" + "=" * 78)
        print(" BACKPROPAGATION NEURAL NETWORK PREDICTION & FEATURE ANALYSIS SYSTEM")
        print("=" * 78)
        print(" 1. Tumor Classification using BPNN")
        print(" 2. Prediction Confidence Analyzer")
        print(" 3. Feature Contribution Analyzer")
        print(" 4. Hidden Neuron Response Analyzer")
        print(" 5. Input Feature Masking Challenge")
        print(" 6. Misclassification Confidence Analyzer")
        print(" 7. BPNN-MLP-SHLNN Comparative Analysis")
        print(" 8. Exit")
        try:
            choice = input("Enter your choice (1-8): ").strip()
        except EOFError:
            break
        if choice == "8":
            print("Goodbye!")
            break
        if choice in actions:
            actions[choice]()
        else:
            print("Invalid choice. Please enter a number from 1 to 8.")


if __name__ == "__main__":
    main()
```
